## 5. С помощью groupby() и агрегирующих функций определим:
### 5.1 какова средняя конверсия (target) в зависимости от возрастной группы

```python
# создание категорий с помощью pd.cut
bins = [18, 26, 36, 46, 56, 66]
labels = ['18-25', '26-35', '36-45', '46-55', '56-65']
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=False)
```

```python
# groupby с агрегацией
age_target = df.groupby(['age_group'], observed=True).agg({'target': ['mean']})
print(age_target)
```

### 5.2 как средний доход и средний показатель вовлечённости различаются между конвертированными и неконвертированными пользователями

```python
# groupby с агрегацией
target_IS = df.groupby(['target'], observed=True).agg({
    'income': ['mean'],
    'score': ['mean']
})
print(target_IS)
```

```python
# определим разницу
d_income = 43446.248276 - 42019.469014
d_score = 0.494552 - 0.463424
print(f'У конвертированных пользователей средний доход больше на {d_income}, а средний показатель вовлечённости выше на {d_score}')
```

### 5.3 как количество посещений связано с уровнем вовлечённости

```python
# на основании статистики из пункта 4 создадим следующие категории посещаемости
bins = [0, 7, 11, 22]
labels = ['Низкое', 'Среднее', 'Высокое']
df['visitor_level'] = pd.cut(df['visits'], bins=bins, labels=labels, right=True)
```

```python
# groupby с агрегацией
visitlvl_S = df.groupby(['visitor_level'], observed=True).agg({'score': ['mean']})
print(visitlvl_S)
```

Посмотрев на результат, можно заметить, что наибольший прирост уровня вовлечённости происходит при переходе с низкой на среднюю частоту посещений аналитика в процентах представлена ниже

```python
print('Прирост вовлечённости при переходе с низкого на средний уровень посещений:', 0.488541/0.419870*100-100, '%')
print('Прирост вовлечённости при переходе со среднего на высокий уровень посещений:', 0.515922/0.488541*100-100, '%')
```

### 5.4 будут ли ощутимые различия между группами с конверсией 1 и конверсией 0

### 5.5 в какой возрастной группе будет максимальная доля конверсии

## 6. С помощью pivot_table() построим:

### 6.1 сводную таблицу с возрастными группами, уровнями дохода и средней конверсией

```python
# создадим категории дохода, опираясь на данные раздела 4
bins = [8000, 31268, 51475, 112113]
labels = ['Низкий', 'Средний', 'Высокий']
df['income_level'] = pd.cut(df['income'], bins=bins, labels=labels, right=True)
```

```python
# pivot_table
pivot = df.pivot_table(
    index='age_group',
    columns='income_level',
    values='target',
    aggfunc='mean',
    observed=False)
print (pivot)
```

Анализируя результаты можно сказать, что в предложенном наборе данных отсутствуют пользователи старше 45 лет, указавшие у себя низкий доход. Также отсутствуют пользователи до 26 лет, указавшие у себя высокий доход. Наивысший средний доход наблюдается у возрастных групп 26-35 и 56-65, низший средний доход у возрастной группы 26-35.

### 6.2 профиль идеального пользователя, основываясь на возрасте, доходе и показателе вовлечённости

```python
# на основании статистики из пункта 4 создадим следующие категории вовлечённости
bins = [0, 0.36, 0.58, 0.87]
labels = ['Низкая', 'Средняя', 'Высокая']
df['score_level'] = pd.cut(df['score'], bins=bins, labels=labels, right=True)
```

```python
# pivot_table
pivot = df.pivot_table(
    index='score_level',
    columns='target',
    values=['age', 'income'],
    aggfunc='mean',
    observed=False)
print (pivot)
```

Опираясь на данные, можно сделать вывод, что под профиль "идеального пользователя" попадают аккаунты с возрастом 33 года или с доходом 42900 у.е.

### 6.3 сводную таблицу с двумя измерениями: score и target

```python
# pivot_table
pivot = df.pivot_table(
    index='score_level',
    columns='target',
    values=['income'],
    aggfunc='mean',
    observed=False)
print (pivot)
```

Исходя из данных таблицы, наивысший средний доход у пользователей с низкой степенью вовлечённости и конверсией равной 1, в то время как низший - у пользователей с высокой степенью вовлечённости и конверсией равной 0. При том важно подметить, что самый крупный рост среднего дохода происходит при переходе со среднего на низкий уровень вовлечённости пользователей вне зависимости от конверсии, а конверсия добавляет лишь небольшой процент.

```python
# докажем вывод выше и вычислим процент прироста
d_target1 = abs(47489.36/44235.385417*100-100)
d_target2 = abs(42487.754601/41298.034582*100-100)
d_target3 = abs(42849.870130/41225.831325*100-100)
d_score1 = abs(47489.36/42487.754601*100-100)
d_score2 = abs(42487.754601/42849.870130*100-100)
d_score3 = abs(44235.385417/41298.034582*100-100)
d_score4 = abs(41298.034582/41225.831325*100-100)
print(f'Максимальный прирост при снижении активности: {max(d_score1, d_score2, d_score3, d_score4)}%, при конверсии = 1 максимальный прирост: {max(d_target1, d_target2, d_target3)}')
print(f'Минимальный прирост при снижении активности: {min(d_score1, d_score2, d_score3, d_score4)}%, при конверсии = 1 минимальный прирост: {min(d_target1, d_target2, d_target3)}')
```

Проведя рассчёты также можно заметить, что конверсия даёт гарантированный прирост в 2.8%, в то время как снижение уровня вовлечённости минимально приносит очень мало (0.18%).

## 7. Создадим новый признак клики за каждый визит

```python
# новый признак = если количество визитов > 0, мы передаём клики/визиты, иначе записываем 0
df['clicks_per_visit'] = np.where(df['visits'] != 0, df['clicks']/df['visits'], 0)
```
